## 1. Why loops? (8 min)

If we need to inspect ten temperature readings, copying an `if` statement ten times is slow and error-prone. A loop says: *for each item, carry out these steps.*

We will use `for` loops first because they are the safest starting point when the number of repetitions is known.

```python
for minute in range(1, 6):
    print(f'Reading at minute {minute}')
```

```python
m = 0 
for i in range(1, 6):
 m = m + 250 
 print(i, m)
```

`range(1, 6)` creates 1, 2, 3, 4, 5: the final number is excluded.

## 2. Repeated calculations (15 min)

Indent all instructions that should repeat. Here, the same conversion calculation is applied at several times.

```python
for time_min in range(0, 31, 5):
    conversion_percent = 100 * (1 - 0.9**time_min)
    print(f'{time_min:2d} min → conversion = {conversion_percent:6.2f}%')
```

## 3. `while` loops: repeat until a condition changes (12 min)

A `while` loop repeats as long as its condition remains true. Make sure something inside the loop changes; otherwise the loop can run forever. For early learning, prefer `for` when possible.

```python
tank_level_percent = 100

while tank_level_percent > 20:
    print(f'Current level: {tank_level_percent}%')
    tank_level_percent = tank_level_percent - 20

print('Low-level threshold reached.')
```

```python
water_level = 100
while water_level > 20:
     print(f'OKAY.')
     water_level = water_level - 20
```

```python
n = 0 
while n <= 10:    
    print (n)   
    n = n + 2
```

```python
n = 10 
while n >= 0:    
    print (n)   
    n = n - 2
```

## 4. Combining loops and decisions (15 min)

A useful pattern is: loop through readings, then use a conditional to label each one. The code below is a simple alarm scan—not a real safety system.

```python
temperatures_c = [72, 84, 96, 88, 101]

for temperature_c in temperatures_c:
    if temperature_c > 100:
        label = 'ALARM'
    elif temperature_c > 85:
        label = 'CAUTION'
    else:
        label = 'normal'
    print(f'{temperature_c} °C → {label}')
```

```python
temperatures_c = [20, 50, 70, 30, 60, 80] 
for temperature_c in temperatures_c:   
     if temperature_c >= 65:
         label = 'Alarm'    
     elif temperature_c > 50:       
         label = 'Near Ok'    
     else:        
         label = 'Normal'   
     print(f'{temperature_c} °C → {label}')
```
